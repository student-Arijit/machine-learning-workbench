```python
try:
  import numpy as np
  import pandas as pd
  import pickle
  from sklearn.model_selection import train_test_split
  from sklearn.ensemble import RandomForestClassifier
  from sklearn.metrics import accuracy_score, confusion_matrix

except ModuleNotFoundError as e:
  print(f"Module not found {e}\nCheck the spelling or use pip install")
```

```python
df = pd.read_csv('/workspaces/machine-learning-workbench/Submarine_mine_detection_system/assets/training_sonar_data.csv', header=None)
df.head()
```

```python
df[60].value_counts()
```

```python
df.groupby(60).mean()
```

```python
x = df.drop(columns=60)
y = df[60]

print(x.head())
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
print(x.shape, X_train.shape, X_test.shape)
print(y.shape, y_train.shape, y_test.shape)
```

```python
model = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42
)
```

```python
model.fit(X_train, y_train)
```

```python
x_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(x_train_prediction, y_train)
```

```python
print(f"Accuracy on training data: {training_data_accuracy*100:.2f}%")
```

```python
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, y_test) 
```

```python
print(f"Accuracy on test data : {test_data_accuracy*100:.2f}%")
```

```python
cm = confusion_matrix(y_test, X_test_prediction)
print(cm)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, X_test_prediction))
```
