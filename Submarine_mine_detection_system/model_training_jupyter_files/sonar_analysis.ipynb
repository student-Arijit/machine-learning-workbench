```python
try:
  import numpy as np
  import pandas as pd
  from sklearn.model_selection import train_test_split
  from sklearn.preprocessing import StandardScaler
  from sklearn.svm import SVC
  from sklearn.pipeline import Pipeline
  from sklearn.model_selection import GridSearchCV
  from sklearn.metrics import accuracy_score

except ModuleNotFoundError as e:
  print(f"Module not found {e}\nCheck the spelling or use pip install")
```

```python
df = pd.read_csv('/workspaces/machine-learning-workbench/Submarine_mine_detection_system/assets/training_sonar_data.csv', header=None)
df.head()
```

```python
df[60].value_counts()
```

```python
df.groupby(60).mean()
```

```python
x = df.drop(columns=60)
y = df[60]

print(x.head())
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
print(x.shape, X_train.shape, X_test.shape)
print(y.shape, y_train.shape, y_test.shape)
```

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC())
])
```

```python
param_grid = {
    'svm__kernel': ['rbf'],
    'svm__C': [0.1, 1, 10, 100],
    'svm__gamma': ['scale', 0.01, 0.001]
}
```

```python
grid = GridSearchCV(pipeline, param_grid, cv=5)
grid.fit(X_train, y_train)
```

```python
print("Best:", grid.best_params_)
print("Accuracy:", grid.score(X_test, y_test))
```

```python
import pickle

best_model = grid.best_estimator_

with open('sonar_model.pkl', 'wb') as f:
    pickle.dump(best_model, f)
```

```python
a = np.asarray(df.iloc[2].drop(60))
a = a.reshape(1, -1)
a.shape
```

```python
prediction = best_model.predict(a)
print(a)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_train, prediction))
```
